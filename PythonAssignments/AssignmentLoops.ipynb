##Program to check if a number is positive, negative or zero

```python
num = int(input("Enter number "))
if num == 0:
  print("Number is zero")
elif num > 0:
  print("Number is positive")
else:
  print("Number is negative")
```

##Program that takes age and classifies them based on the age

```python
age = int(input("Enter age "))
if age < 0:
  print("Invalid age")
elif age < 13:
  print("Child")
elif age >=13 and age <=19:
  print("Teenager")
elif age >=20 and age <=59:
  print("Adult")
else:
  print("Senior")
```

##Program to check if character entered by user is vowel or consonant

```python
character = input("Enter character ")
character = character.lower()
if character == 'a' or character == 'e' or character == 'i' or character == 'o' or character == 'u':
  print("Vowel")
else:
  print("Consonants")
```

##Program to check if number is divisible by 3 and 5

```python
num = int(input("Enter number "))
if num % 3 == 0 and num % 5 == 0:
  print("Number is divisible by 3 and 5")
else:
  print("Number is not divisible")
```

## Program to print sum of first n natural numbers

```python
count = int(input("Enter count of numbers "))
sum = 0
while count > 0:
  sum = sum + count
  count = count - 1
print(sum)
```

##Program to print multiplication table of a given number upto 10

```python
num = int(input("Enter number"))
for i in range(1,11):
  print(num,"*",i,"=",num*i)
```

##Program to print factorial of number using loop


```python
num = int(input("Enter number"))
fact = 1
while num > 0:
  fact = fact * num
  num = num-1
print(fact)
```

## Program to reverse a number

```python
num = int(input("Enter number"))
rev_no = 0
while num >0:
  digit = num%10
  rev_no = rev_no*10 + digit
  num = num//10
print(rev_no)
```

##Program to print fibonacci number till n terms

```python
fib0= 0
fib1= 1
count = 0
n = int(input("Enter number of terms"))
if n == 1:
  print(fib0)
else:
  while n > count:
    print(fib0)
    fib2= fib0 + fib1
    fib0 = fib1
    fib1 = fib2
    count+=1
```
