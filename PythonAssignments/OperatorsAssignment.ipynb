# Mathematical Operators
### Program to calculate perimeter and area of rectangle



```python
length = int(input("Enter the length of rectangle"))
breadth = int(input("Enter the breadth of rectangle"))
perimeter = 2*(length+breadth)
area = length*breadth
print("Perimeter of rectangle is", perimeter)
print("Area of rectangle is", area)
```

# Arithmetic Operators
### Program to perform all arithmetic operations


```python
a = int(input("Enter the value of a"))
b = int(input("Enter the value of b"))
print("Addition of a and b is", a+b)
print("Subtraction of a and b is", a-b)
print("Multiplication of a and b is", a*b)
print("Division of a and b is", a/b)
print("Modulus of a and b is", a%b)
print("Exponential value of and b is", a**b)
```

# Operator precedence

```python
a = 10+3*2**2 # exponential has higher precedence then multiplication then addition
print(a)
```

# Conditional Statements
### Program to check odd or even

```python
num = int(input("Enter a number "))
if num%2 == 0:
   print("number is even")
else:
   print("number is odd")
```

### Program to find maximum of three numbers


```python
num1 = int(input("Enter first number "))
num2 = int(input("Enter second number "))
num3 = int(input("Enter third number "))
if num1>num2 and num1>num3:
   print("First number is maximum")
elif num2>num3:
   print("Second number is maximum")
else:
   print("Third number is maximum")
```

### Program to print grades of students wrt marks scored

```python
marks = int(input("Enter marks scored "))
if marks>=90:
   print("Grade A")
elif marks>=75:
   print("Grade B")
elif marks>=50:
   print("Grade C")
elif marks<50:
   print("Grade F")
```

### Program to check if a year is leap year

```python
year = int(input("Enter year "))
if year%4 == 0:
  if year%100 == 0:
    if year%400 == 0:
      print("Year is leap year")
    else:
      print("Year is not leap year")
  else:
    print("Year is leap year")
else:
  print("Year is not leap year")
```

### Program to build simplen calculator

```python
num1 = int(input("Enter num1 "))
num2 = int(input("Enter num2 "))
operator = input("Enter operator ")
if operator == '+':
  res = num1+num2
elif operator == '-':
  res = num1-num2
elif operator == '*':
  res = num1*num2
elif operator == '/':
  res = num1/num2
print("result is ", res)
```
